```
import numpy as np
from matplotlib.pyplot import *
from pylab import *
K = 1000
s = 0.08
i = s
n = 10
t = linspace(0, n, n)
y1 = np.ones(len(t)) * K # a horizontal line
y2 = K * (1 + s * t) # a(t) = (1+s*t) ; A(t) = K * (1+s*t)
y3 = K * (1 + i)**t # a(t) = ( 1 + i)**t ; A(t) = K * ( 1 + i)**t
title("Simple Interest Vs Compound Interest")
xlabel('Number of years')
ylabel('Values')
xlim(0,11)
ylim(800, 2200)
plot(t, y1, 'b-')
plot(t, y2, 'g--')
plot(t, y3, 'r-')
show()
```

Nominal Rate, Effective Rate, Discount Rate \

APR - Annual Percentage Rate: $i_{nom} = i^{(m)}$ \

EAR - Effective Annual Rate: $i$ \

$ i^{(m)} = m*((1+i)^{(1/m)}-1)$ \

$ i = (1+\frac{i^{(m)}}{m})^m -1$ \

$ d^{(m)} = \frac{i^{(m)}}{1+\frac{i^{(m)}}{m}}$

```
import math
def i_nomi(i, m):
  i_nom = (((1+i)**(1/m))-1)*m
  d_nom = i_nom / (1 + (i_nom/m))
  i_eff = i
  d_eff = i / (1 + i)
  delta = math.log(1+i)
  return i_nom, d_nom, i_eff, d_eff, delta
```

```
print(i_nomi(0.12, 2)[0])
print(i_nomi(0.12, 2)[1])
print(i_nomi(0.12, 2)[2])
print(i_nomi(0.12, 2)[3])
print(i_nomi(0.12, 2)[4])
```

```
from sympy import symbols
from sympy import plot
t = symbols('t')
x = (1.2)**t
plot(x, (t,0,10), ylabel="a(t)")
plt.figure(figsize=(10,10))
plt.show()
```

```
from sympy import symbols
from numpy import linspace
from sympy import lambdify
import matplotlib.pyplot as mpl
t = symbols('t')
x = (1.2)**t
lam_x = lambdify(t, x, modules=['numpy'])
x_vals = linspace(0, 6, 6)
y_vals = lam_x(x_vals)
mpl.plot(x_vals, y_vals, marker = 'o')
mpl.ylabel("a(t)")
mpl.show()
```
